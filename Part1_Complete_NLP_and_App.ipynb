## 1. Environment & System Logging Initialization

```python
import concurrent.futures
import logging
import os
from typing import Tuple

logging.basicConfig(
    level=logging.INFO,
    format="%(asctime)s [%(levelname)s] %(name)s - %(message)s"
)
logger = logging.getLogger("Part1_Financial_NLP_Suite")
print("System logging initialized successfully for Part1_Financial_NLP_Suite.")
```

## 2. Defensive Validation & Multi-Task Sub-Task Definitions

```python
def validate_inputs(api_key: str, text_content: str, use_mock: bool = False) -> Tuple[bool, str]:
    if not text_content or len(text_content.strip()) < 20:
        return False, "Input payload too short (minimum 20 characters required)."
    if use_mock:
        return True, ""
    if not api_key or not api_key.strip():
        return False, "OpenAI API Key is required."
    if not api_key.strip().startswith("sk-"):
        return False, "Invalid API Key format (must start with 'sk-')."
    return True, ""

def run_subtask_1(llm, text: str, use_mock: bool = False) -> str:
    if use_mock or llm is None:
        return (
            "Summary:\n"
            "- Revenue & Growth: Revenue declined 4% YoY to $4.2B driven by macro disruptions.\n"
            "- Debt & Balance Sheet: Net debt rose to $1.8B post-acquisition; Debt-to-EBITDA reached 3.8x.\n"
            "- Guidance & Outlook: Full-year EBITDA guidance reduced by 150 bps due to supply chain friction."
        )
    from langchain_core.prompts import ChatPromptTemplate
    prompt = ChatPromptTemplate.from_template("Summarize financial metrics for:\n{input}")
    return (prompt | llm).invoke({"input": text}).content

def run_subtask_2(llm, text: str, use_mock: bool = False) -> str:
    if use_mock or llm is None:
        return (
            "Risk Triage Level: High\n"
            "Market Sentiment: Bearish\n"
            "Domain Metrics Assessment:\n"
            "- Debt-to-EBITDA Ratio Concern: 3.8x ratio exceeds prudent target threshold of 3.0x.\n"
            "- Liquidity Status ($ Cash): Stable cash balance of $600M provides short-term buffer.\n"
            "- Credit Risk Indicators: Leverage expansion and guidance markdown increase default sensitivity.\n"
            "- Covenant Breach Warning: Yes - High concern regarding maximum leverage covenant ceilings."
        )
    from langchain_core.prompts import ChatPromptTemplate
    prompt = ChatPromptTemplate.from_template("Evaluate credit risk indicators for:\n{input}")
    return (prompt | llm).invoke({"input": text}).content

def run_subtask_3(llm, text: str, use_mock: bool = False) -> str:
    if use_mock or llm is None:
        return (
            "Recommended Actions:\n"
            "1. Action 1 (Covenant Compliance Audit): Issue formal compliance certificate request to verify 3.8x ratio against limits.\n"
            "2. Action 2 (Liquidity & Leverage Mitigation): Require borrower to present 12-month debt de-leveraging plan.\n"
            "3. Action 3 (Watchlist & Underwriting Review): Move corporate borrower to Credit Watchlist Grade B."
        )
    from langchain_core.prompts import ChatPromptTemplate
    prompt = ChatPromptTemplate.from_template("Extract top 3 covenant audit actions for:\n{input}")
    return (prompt | llm).invoke({"input": text}).content

print("All 3 financial NLP sub-task chains and input validators registered.")
```

## 3. Multi-Threaded Concurrent Execution Verification (`ThreadPoolExecutor`)

```python
sample_financial_payload = (
    "Company XYZ reported Q3 revenue of $4.2B, down 4% YoY. Net debt increased to $1.8B following the acquisition, "
    "raising Debt-to-EBITDA to 3.8x. Management lowered full-year EBITDA guidance by 150 bps due to supply chain disruptions "
    "and rising interest expenses. Liquidity remains stable with $600M cash."
)

use_mock = True
is_valid, validation_msg = validate_inputs("", sample_financial_payload, use_mock=use_mock)
assert is_valid, validation_msg

print(f"Executing 3 sub-tasks concurrently (Mock Mode = {use_mock})...")
with concurrent.futures.ThreadPoolExecutor(max_workers=3) as executor:
    f1 = executor.submit(run_subtask_1, None, sample_financial_payload, use_mock)
    f2 = executor.submit(run_subtask_2, None, sample_financial_payload, use_mock)
    f3 = executor.submit(run_subtask_3, None, sample_financial_payload, use_mock)

    print("\n--- 1. EXECUTIVE SUMMARY ---")
    print(f1.result())
    print("\n--- 2. CREDIT RISK METRICS ---")
    print(f2.result())
    print("\n--- 3. COVENANT ACTIONS ---")
    print(f3.result())
```
