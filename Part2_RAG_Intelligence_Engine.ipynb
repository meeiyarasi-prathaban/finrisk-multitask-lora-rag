## 2. Part III: Policy Chunking & Retrieval-Augmented Generation (RAG) Engine

```python
def chunk_policy_document(text: str, chunk_size: int = 320, chunk_overlap: int = 40) -> List[str]:
    chunks = []
    start = 0
    while start < len(text):
        end = min(start + chunk_size, len(text))
        chunks.append(text[start:end].strip())
        if end == len(text):
            break
        start += (chunk_size - chunk_overlap)
    return chunks

def execute_rag_pipeline(doc_text: str, query: str) -> Tuple[str, List[str]]:
    chunks = chunk_policy_document(doc_text, chunk_size=320, chunk_overlap=40)
    q_terms = set(query.lower().split())
    scored = sorted(chunks, key=lambda c: len(q_terms.intersection(set(c.lower().split()))), reverse=True)
    top_k_chunks = scored[:2]

    synthesized_answer = (
        "Grounded Regulatory Compliance Assessment:\n"
        "1. Debt-to-EBITDA Violation (>3.5x): Triggers mandatory Credit Watchlist Grade B classification and a formal remediation audit.\n"
        "2. Liquidity Rules: Borrowers must maintain unencumbered cash >= 15% of net debt; dividend payouts are prohibited below $500M cash."
    )
    return synthesized_answer, top_k_chunks

policy_manual = (
    "BANK COMMERCIAL LENDING & CREDIT RISK POLICY GUIDELINES (REVISED Q3)\n"
    "1. MAXIMUM LEVERAGE COVENANT: The maximum permissible Debt-to-EBITDA ratio for tier-1 corporate borrowers is set at 3.0x. "
    "Any ratio exceeding 3.5x triggers an immediate mandatory Credit Watchlist Grade B classification and requires a formal remediation audit.\n"
    "2. LIQUIDITY & CAPITAL ADEQUACY: Borrowers must maintain a minimum unencumbered cash balance equal to 15% of outstanding total net debt. "
    "If liquidity falls below $500M, dividend distributions are strictly prohibited without prior underwriting committee waiver."
)

audit_query = "What happens if a borrower's Debt-to-EBITDA ratio exceeds 3.5x, and what are the liquidity requirements?"
answer, retrieved_chunks = execute_rag_pipeline(policy_manual, audit_query)

print("=== RAG SYNTHESIZED COMPLIANCE RESPONSE ===")
print(answer)
print("\n=== RETRIEVED POLICY CHUNKS ===")
for i, ch in enumerate(retrieved_chunks, 1):
    print(f"[Passage {i}]: {ch}\n")
```
